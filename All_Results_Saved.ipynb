```python
import json
import pandas as pd
import os

BACKUP_DIR = '/content/drive/MyDrive/thesis_backup_day1'

os.makedirs(BACKUP_DIR, exist_ok=True)

##All experiment results
results = {
    'Baseline': {
        'ssl': False,
        'rebalance': False,
        'aug': False,
        'acc': 0.721,
        'mel_recall': 0.0,
        'bkl_recall': 0.0,
        'nv_recall': 1.0
    },
    'SSL_Only': {
        'ssl': True,
        'rebalance': False,
        'aug': False,
        'acc': 0.721,
        'mel_recall': 0.0,
        'bkl_recall': 0.05,
        'nv_recall': 0.99
    },
    'Rebalance_Only': {
        'ssl': False,
        'rebalance': True,
        'aug': True,
        'acc': 0.644,
        'mel_recall': 0.375,
        'bkl_recall': 0.62,
        'nv_recall': 0.68
    },
    'SSL_Rebalance': {
        'ssl': True,
        'rebalance': True,
        'aug': True,
        'acc': 0.673,
        'mel_recall': 0.50,
        'bkl_recall': 0.14,
        'nv_recall': 0.84
    },
}

## Created DataFrame for thesis table
df = pd.DataFrame(results).T
df = df[['ssl', 'rebalance', 'aug', 'acc', 'mel_recall', 'bkl_recall', 'nv_recall']]

print("="*32)
print("FINAL ABLATION TABLE FOR THESIS")
print("="*32)
print(df.to_string())
print("="*32)

## Save
summary = {
    'experiments': results,
    'key_finding': 'SSL and rebalancing are both necessary. Neither alone achieves target MEL recall.',
    'clinical_significance': '50% melanoma recall vs 0% baseline = 4 additional melanomas detected per 8 cases',
    'limitation': 'SSL+Rebalance accuracy (67.3%) lower than baseline (72.1%) due to minority-class focus'
}

output_path = f'{BACKUP_DIR}/final_thesis_summary.json'
with open(output_path, 'w') as f:
    json.dump(summary, f, indent=2)

print(f"\n Saved to: {output_path}")
```
